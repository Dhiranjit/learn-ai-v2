# Tool Call

```python
from learn_ai.llm.client import LLMClient
from learn_ai.llm.tools import TOOLS
import json
```

```python
client = LLMClient(
    provider="groq",
    model="qwen/qwen3-32b"
)
```

```python
messages = [
    {"role": "user", "content": "Hi, how are you???"}
]
```

```python
out = client.chat(messages, tools=TOOLS)
out
```

```python
out.tool_calls[0].function.name
```

```python
out.tool_calls[0].function.arguments
```

```python
json.loads(out.tool_calls[0].function.arguments)
```
